```julia
#1.
function lsqn(A,b)
    B=A'*A
    bhat=A'*b
    R= cholesky(B).U
    y=R'\bhat
    x=R\y
    return x
end
```

```julia
A=[1 3 1;1 3 7;1 -1 -4;1 -1 2]
b=[1 1 1 1]'
```

```julia
lsqn(A,b)
```

```julia
function qrfun(A,b)
    F=qr(A)
    Q=Matrix(F.Q)
    R=F.R
    c=Q'*b
    x=R\c
    return x
end
```

```julia
A=[1 3 1;1 3 7;1 -1 -4;1 -1 2]
b=[1 1 1 1]'
```

```julia
qrfun(A,b)
```

```julia
function solve_qr(A::Matrix, b::Vector)
    F = qr(A)
    Q = Matrix(F.Q)
    R = F.R
    Qtb = Q' * b
    
    # Step 4: Solve Rx = Qᵀb
    x = R \ Qtb
    
    # Step 5: Display results
    println("Q = ")
    println(Q)
    println("\nR = ")
    println(R)
    println("\nQᵀb = ")
    println(Qtb)
    println("\nSolution x = ")
    println(x)
    
    return x
end

# Example usage:
A = [1 3 1;
     1 3 7;
     1 -1 -4;
     1 -1 2]

b = [1, 1, 1, 1]

# Call the function
x = solve_qr(A, b)
```

```julia
using LinearAlgebra
function chole(A,b)
    F=cholesky(A)
    R=F.U
    y=R'\b
    x=R\y
    return x
end
    
```

```julia
A=[2 -1 0;-1 2 -1;0 -1 2]
b=[1 1 1]'
```

```julia
isposdef(A)
```

```julia
chole(A,b)
```

```julia
function lsn(A,b)
    B=A'*A
    R=cholesky(B).U
    bhat=A'*b
    y=R'\bhat
    x=R\y
    return x
end
    
```

```julia
A=[1 3 1;1 3 7;1 -1 -4;1 -1 2]
b=[1 1 1 1]'
```

```julia
lsn(A,b)
```

```julia
using LinearAlgebra
function qrd(A, b)
    F=qr(A)
    Q=Matrix(F.Q)
    R=F.R
    C=Q'*b
    x=R\C
    return x
end
    
```

```julia
A=[1 3 1;1 3 7;1 -1 -4;1 -1 2]
b=[1 1 1 1]'
```

```julia
qrd(A,b)
```
