**Exercise 2**

Let $$A = \begin{pmatrix}2&-1\\0&1\\2&2\end{pmatrix}, \quad b=\begin{pmatrix}1\\-5\\6\end{pmatrix}\quad \text{such that}\quad Ax=b,\quad x\in\mathbb{R}^2  \label{eq2}\tag{2}$$

(a) Explain the different steps to solve problem \eqref{eq2} above.

```julia
function chole(A,b)
    B=A'*A
    bhat=A'*b
    R=cholesky(B).U
    y=R'\bhat
    x=R\y
    return x
end
```

```julia
A=[2 -1;0 1;2 2]
b=[1 -5 6]'
```

```julia
chole(A,b)
```

(b) Use your forward and backward function code to solve the problem (2) above.

```julia
function forwad(A,b)
    B=A'*A
    bhat=A'*b
    R=cholesky(B).U
    y=R'*bhat
    return y
end
        
```

```julia
function backword(A,b)
    B=A'*A
    bhat=A'*b
    R=cholesky(B).U
    y=R'*bhat
    x=R*y
    return x
end
```

```julia
backword(A,b)
```

```julia
forwad(A,b)
```
