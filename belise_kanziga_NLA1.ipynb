### Q5.

```julia
using LinearAlgebra
using Random

function matvec_loops(A, x)
    n = size(A, 1)
    b = zeros(n)
    for j in 1:n
        for i in 1:n
            b[i] = b[i] + A[i,j] * x[j]
        end
    end
    return b
end

# Experiment with different sizes
sizes = [3, 10, 100]

println("Timing matrix-vector multiplication:")
println("Size | Built-in | Loops")
println("-----|----------|------")

for n in sizes
    A = randn(n, n)
    x = randn(n)
    
    # Warm up
    A*x
    matvec_loops(A, x)
    
    # Time built-in
    t1 = @elapsed A*x
    # Time loops
    t2 = @elapsed matvec_loops(A, x)
    
    println("$n×$n | $t1 s | $t2 s")
end

```

#### Built-in multiplication will be faster due to optimized BLAS routines.
