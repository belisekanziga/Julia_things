```julia
using LinearAlgebra
function decomp(A,b)
    F=qr(A)
    Q=collect(F.Q)
    R1=F.R
    R=[R1;0 0 0]
    c=Q'*b
    x=R\c
    return x
end
    
```

```julia
A=[1 3 1;1 3 7;1 -1 -4;1 -1 2]
b=[1 1 1 1]'
```

```julia
decomp(A,b)
```
